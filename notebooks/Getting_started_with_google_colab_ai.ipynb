```python
import matplotlib.pyplot as plt
import numpy as np

# Số liệu cho Đồ thị 1
log_H = np.array([0.21, 0.22, 0.29, 0.39])
log_Q = np.array([2.53, 2.66, 2.81, 2.98])

# Số liệu cho Đồ thị 2
H = np.array([1.64, 1.67, 1.95, 2.48])
Cd_TN = np.array([0.538, 0.719, 0.791, 0.822])
Cd_Tsu = np.array([0.614, 0.614, 0.616, 0.620])

# Cài đặt kích thước figure cho 2 đồ thị
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

# --- VẼ ĐỒ THỊ 1: log(Q) = f[log(H)] ---
ax1.plot(log_H, log_Q, marker='o', linestyle='-', color='b', label='Dữ liệu đo')
# Thêm đường xu hướng (Trendline)
z = np.polyfit(log_H, log_Q, 1)
p = np.poly1d(z)
ax1.plot(log_H, p(log_H), "r--", label=f'Đường xu hướng: y = {z[0]:.2f}x + {z[1]:.2f}')

ax1.set_title('Hình H3. Đường quan hệ log(Q) = f[log(H)]')
ax1.set_xlabel('log(H)')
ax1.set_ylabel('log(Q)')
ax1.grid(True, which='both', linestyle='--', linewidth=0.5)
ax1.legend()

# --- VẼ ĐỒ THỊ 2: Cd = f(H) ---
ax2.plot(H, Cd_TN, marker='o', linestyle='-', color='r', label='Cd Thí nghiệm')
ax2.plot(H, Cd_Tsu, marker='s', linestyle='--', color='g', label='Cd Tsugaev')

ax2.set_title('Hình H4. Đường quan hệ Cd = f(H)')
ax2.set_xlabel('H (cm)')
ax2.set_ylabel('Hệ số lưu lượng Cd')
ax2.grid(True, which='both', linestyle='--', linewidth=0.5)
ax2.legend()

# Hiển thị đồ thị
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.ticker import AutoMinorLocator

# --- SỐ LIỆU ĐÃ NHẬP TỪ BẢNG CỦA BẠN ---
# Hình H.3
log_H = np.array([0.21, 0.22, 0.29, 0.39])
log_Q = np.array([2.53, 2.66, 2.81, 2.98])

# Hình H.4
H = np.array([1.64, 1.67, 1.95, 2.48])
Cd_TN = np.array([0.538, 0.719, 0.791, 0.822])
Cd_Tsu = np.array([0.614, 0.614, 0.616, 0.620])

# Cài đặt phông chữ và kích thước tổng thể
plt.rcParams.update({'font.size': 11})
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 7))

# ==========================================
# ĐỒ THỊ 1: Hình H3. log(Q) = f[log(H)]
# ==========================================
# Vẽ điểm dữ liệu thực tế
ax1.plot(log_H, log_Q, marker='o', markersize=8, linestyle='', color='blue', label='Số liệu thí nghiệm')

# Hồi quy tuyến tính để tìm b (giao điểm trục tung) và hệ số góc
z = np.polyfit(log_H, log_Q, 1)
p = np.poly1d(z)
he_so_goc = z[0]
giao_diem_b = z[1] # Đây chính là giá trị b bạn cần tìm trong phuong trình

# Vẽ đường xu hướng
x_trend = np.linspace(0.15, 0.45, 100)
ax1.plot(x_trend, p(x_trend), "r-", linewidth=2, label=f'Đường bậc nhất: y = {he_so_goc:.2f}x + {giao_diem_b:.2f}\n(Giao điểm b = {giao_diem_b:.2f})')

# Căn chỉnh trục & Giấy ô ly
ax1.set_title('H3. Đường quan hệ log(Q) = f[log(H)] của bờ tràn chữ nhật', fontweight='bold', pad=15)
ax1.set_xlabel('log(H), cm')
ax1.set_ylabel('log(Q), $cm^3/s$')
ax1.set_xlim(0.15, 0.45)
ax1.set_ylim(2.4, 3.1)

# Chia vạch chi tiết để tạo ô ly
ax1.xaxis.set_minor_locator(AutoMinorLocator(5))
ax1.yaxis.set_minor_locator(AutoMinorLocator(5))
ax1.grid(which='major', color='#666666', linestyle='-', linewidth=1.2)
ax1.grid(which='minor', color='#999999', linestyle='-', linewidth=0.4)
ax1.legend(loc='lower right', framealpha=1, edgecolor='black')


# ==========================================
# ĐỒ THỊ 2: Hình H4. Cd = f(H)
# ==========================================
# Vẽ 2 đường đồ thị
ax2.plot(H, Cd_TN, marker='o', markersize=8, linestyle='-', linewidth=2, color='red', label='Cd Thí nghiệm')
ax2.plot(H, Cd_Tsu, marker='s', markersize=8, linestyle='--', linewidth=2, color='green', label='Cd Tsugaev')

# Căn chỉnh trục & Giấy ô ly
ax2.set_title('H4. Đường quan hệ Cd = f(H) của bờ tràn chữ nhật', fontweight='bold', pad=15)
ax2.set_xlabel('H, cm')
ax2.set_ylabel('$C_d$')
ax2.set_xlim(1.5, 2.6)
ax2.set_ylim(0.5, 0.9)

# Chia vạch chi tiết để tạo ô ly
ax2.xaxis.set_minor_locator(AutoMinorLocator(5))
ax2.yaxis.set_minor_locator(AutoMinorLocator(5))
ax2.grid(which='major', color='#666666', linestyle='-', linewidth=1.2)
ax2.grid(which='minor', color='#999999', linestyle='-', linewidth=0.4)
ax2.legend(loc='lower right', framealpha=1, edgecolor='black')

# Hiển thị
plt.tight_layout(pad=3.0)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.ticker import MultipleLocator

# ==========================================
# 1. NHẬP SỐ LIỆU TỪ BẢNG CỦA BẠN
# ==========================================
# Số liệu Hình H.3
log_H = np.array([0.21, 0.22, 0.29, 0.39])
log_Q = np.array([2.53, 2.66, 2.81, 2.98])

# Số liệu Hình H.4
H = np.array([1.64, 1.67, 1.95, 2.48])
Cd_TN = np.array([0.538, 0.719, 0.791, 0.822])
Cd_Tsu = np.array([0.614, 0.614, 0.616, 0.620])

# ==========================================
# 2. KHỞI TẠO ĐỒ THỊ
# ==========================================
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 8))
# Tùy chỉnh font chữ để giống báo cáo học thuật
plt.rcParams.update({'font.family': 'serif', 'font.size': 12})

# Cài đặt kiểu khung nhãn (Text Box) cho giống hệt file PDF
bbox_props = dict(boxstyle="square,pad=0.4", fc="white", ec="black", lw=1.5)

# ==========================================
# 3. VẼ HÌNH H.3: log(Q) = f[log(H)]
# ==========================================
# Giới hạn trục để tạo khung ô ly vuông vức
ax1.set_xlim(0.15, 0.45)
ax1.set_ylim(2.4, 3.1)

# Chia vạch để tạo lưới ô ly milimet (10 ô nhỏ trong 1 ô lớn)
ax1.xaxis.set_major_locator(MultipleLocator(0.05))
ax1.xaxis.set_minor_locator(MultipleLocator(0.005))
ax1.yaxis.set_major_locator(MultipleLocator(0.1))
ax1.yaxis.set_minor_locator(MultipleLocator(0.01))

# Kẻ lưới
ax1.grid(which='major', color='black', linewidth=1.2)
ax1.grid(which='minor', color='#555555', linewidth=0.5)

# Đặt nhãn vào các góc y hệt PDF
ax1.text(0.01, 0.99, 'logQ (cm$^3$/s)', transform=ax1.transAxes, verticalalignment='top', bbox=bbox_props, fontweight='bold')
ax1.text(0.99, 0.01, 'logH, cm', transform=ax1.transAxes, horizontalalignment='right', verticalalignment='bottom', bbox=bbox_props, fontweight='bold')

# Hồi quy tuyến tính để tìm phương trình y = ax + b
z = np.polyfit(log_H, log_Q, 1)
p = np.poly1d(z)
he_so_goc = z[0]
giao_diem_b = z[1]

# Vẽ đường thẳng và các điểm dữ liệu
x_trend = np.linspace(0.15, 0.45, 100)
ax1.plot(x_trend, p(x_trend), "r-", linewidth=2.5, label=f'Đường quan hệ (b = {giao_diem_b:.3f})')
ax1.plot(log_H, log_Q, 'ko', markersize=8, markerfacecolor='blue', label='Số liệu thực nghiệm')

ax1.set_title('H3. Đường quan hệ log(Q) = f[log(H)]', pad=20, fontweight='bold')
ax1.legend(loc='lower right', bbox_to_anchor=(0.95, 0.15), framealpha=1, edgecolor='black')

# ==========================================
# 4. VẼ HÌNH H.4: Cd = f(H)
# ==========================================
# Giới hạn trục để tạo khung ô ly
ax2.set_xlim(1.5, 2.6)
ax2.set_ylim(0.5, 0.9)

# Chia vạch lưới ô ly milimet
ax2.xaxis.set_major_locator(MultipleLocator(0.1))
ax2.xaxis.set_minor_locator(MultipleLocator(0.01))
ax2.yaxis.set_major_locator(MultipleLocator(0.05))
ax2.yaxis.set_minor_locator(MultipleLocator(0.005))

# Kẻ lưới
ax2.grid(which='major', color='black', linewidth=1.2)
ax2.grid(which='minor', color='#555555', linewidth=0.5)

# Đặt nhãn vào các góc y hệt PDF
ax2.text(0.01, 0.99, 'C$_d$', transform=ax2.transAxes, verticalalignment='top', bbox=bbox_props, fontweight='bold', fontsize=14)
ax2.text(0.99, 0.01, 'H, cm', transform=ax2.transAxes, horizontalalignment='right', verticalalignment='bottom', bbox=bbox_props, fontweight='bold')

# Vẽ dữ liệu
ax2.plot(H, Cd_TN, marker='o', markersize=8, linestyle='-', linewidth=2, color='red', markerfacecolor='white', markeredgewidth=2, label='Cd Thí nghiệm')
ax2.plot(H, Cd_Tsu, marker='s', markersize=8, linestyle='--', linewidth=2, color='green', markerfacecolor='white', markeredgewidth=2, label='Cd Tsugaev')

ax2.set_title('H4. Đường quan hệ Cd = f(H)', pad=20, fontweight='bold')
ax2.legend(loc='lower right', bbox_to_anchor=(0.95, 0.15), framealpha=1, edgecolor='black')

# ==========================================
# 5. XUẤT ĐỒ THỊ
# ==========================================
# Tắt các nhãn trục ngoài lề để tập trung vào hộp nhãn bên trong giống mẫu
ax1.tick_params(axis='both', which='both', labelbottom=False, labelleft=False)
ax2.tick_params(axis='both', which='both', labelbottom=False, labelleft=False)

plt.tight_layout(pad=3.0)
plt.show()

# In kết quả giao điểm b ra màn hình để điền vào báo cáo
print("-" * 50)
print(f"KẾT QUẢ ĐIỀN VÀO BÁO CÁO (MỤC III.B.1):")
print(f"Giao điểm của đồ thị với trục tung, b = {giao_diem_b:.4f}")
print("-" * 50)
```
